### Basic Data Quality Checks

```python
def summarize_df(df, name):
    print(f"-- {name} --")
    print(f"Shape: {df.shape}")
    display(df.sample(min(3, len(df))))
    nulls = df.isna().sum()
    if nulls.any():
        print("Missing values (top 20 by count):")
        display(nulls[nulls > 0].sort_values(ascending=False).head(20).to_frame("missing"))
    else:
        print("No missing values")
    print("\n" * 2)

for name, df in datasets.items():
    summarize_df(df, name)
```

### Descriptive Statistics

```python
def add_margin(df):
    """Ensure a Margin column exists (WScore - LScore)."""
    if {"WScore","LScore"}.issubset(df.columns):
        df = df.copy()
        df["Margin"] = df["WScore"] - df["LScore"]
    return df

def summarize_scores_margins(df, label):
    """Return summary stats for winning score, losing score, and margin."""
    return pd.DataFrame({
        "Mean": [df["WScore"].mean(), df["LScore"].mean(), df["Margin"].mean()],
        "Median": [df["WScore"].median(), df["LScore"].median(), df["Margin"].median()],
        "Std": [df["WScore"].std(), df["LScore"].std(), df["Margin"].std()]
    }, index=[f"{label} - Winning Score",f"{label} - Losing Score",f"{label} - Win Margin"]).round(2)

def plot_side_by_side(rs_df, tourney_df, title_prefix=""):
    """Compare regular season vs. tournament score & margin distributions."""
    fig, axes = plt.subplots(2, 2, figsize=(14,10))

    # Scores
    sns.histplot(rs_df["WScore"], bins=40, kde=True, label="Winning", ax=axes[0,0])
    sns.histplot(rs_df["LScore"], bins=40, kde=True, color="red", label="Losing", ax=axes[0,0])
    axes[0,0].legend()
    axes[0,0].set_title(f"{title_prefix}Regular Season — Scores")

    sns.histplot(tourney_df["WScore"], bins=30, kde=True, label="Winning", ax=axes[0,1])
    sns.histplot(tourney_df["LScore"], bins=30, kde=True, color="red", label="Losing", ax=axes[0,1])
    axes[0,1].legend()
    axes[0,1].set_title(f"{title_prefix}NCAA Tournament — Scores")

    # Margins
    sns.histplot(rs_df["Margin"], bins=30, ax=axes[1,0])
    axes[1,0].set_title(f"{title_prefix}Regular Season — Win Margins")

    sns.histplot(tourney_df["Margin"], bins=20, ax=axes[1,1])
    axes[1,1].set_title(f"{title_prefix}NCAA Tournament — Win Margins")

    plt.tight_layout()
    plt.show()

# Prepare data
df_men_rs_compact = add_margin(df_men_rs_compact)
df_women_rs_compact = add_margin(df_women_rs_compact)
df_men_ncaa_compact = add_margin(df_men_ncaa_compact)
df_women_ncaa_compact = add_margin(df_women_ncaa_compact)
```

```python
# Men’s comparison
plot_side_by_side(df_men_rs_compact, df_men_ncaa_compact, "Men — ")
```

```python
# Women’s comparison
plot_side_by_side(df_women_rs_compact, df_women_ncaa_compact, "Women — ")
```

```python
print("-- Men --")
men_summary = pd.concat([
    summarize_scores_margins(df_men_rs_compact, "Men Regular Season"),
    summarize_scores_margins(df_men_ncaa_compact, "Men NCAA Tournament")
])
display(men_summary)

print("-- Women --")
women_summary = pd.concat([
    summarize_scores_margins(df_women_rs_compact, "Women Regular Season"),
    summarize_scores_margins(df_women_ncaa_compact, "Women NCAA Tournament")
])
display(women_summary)
```

* Women in general have larger win margins than men during the regular season
* Women have higher and men have lower win margin in NCAA tournament --> check for rounds
* Women have higher winning and losing score in NCAA tournament compared to regular season
* Men have exactly the same winning and losing score (median) in NCAA tournament compared to regular season
* Points scored distributions during regular season follow a clean normal distribution for both men and women

#### NCAA Tournament Round Analysis

```python
def add_rounds(tourney_df):
    """Add simple round groupings based on DayNum thresholds.
    Note: this is approximate, since NCAA brackets can vary.
    """
    df = tourney_df.copy()
    # Tournament usually starts around DayNum 135–137
    df["Round"] = pd.cut(
        df["DayNum"],
        bins=[0, 140, 147, 999],
        labels=["Early Rounds (64/32)", "Middle Rounds (Sweet16/Elite8)", "Final Four/Champ"]
    )
    return df

df_men_ncaa_compact = add_rounds(df_men_ncaa_compact)
df_women_ncaa_compact = add_rounds(df_women_ncaa_compact)

def plot_margin_by_round(df, title_prefix=""):
    plt.figure(figsize=(8,5))
    sns.boxplot(x="Round", y="Margin", data=df, order=["Early Rounds (64/32)","Middle Rounds (Sweet16/Elite8)","Final Four/Champ"])
    plt.title(f"{title_prefix}Tournament Win Margins by Round")
    plt.show()

plot_margin_by_round(df_men_ncaa_compact, "Men — ")
plot_margin_by_round(df_women_ncaa_compact, "Women — ")
```

* Both men and women have higher average win margin in early rounds --> due to best seed against worst seeds
* Women median win margin stays roughly the same across all rounds
* There are only outliers towards high win margins

### Winning/Losing Teams Statistics

```python
def build_long_form(rs_detailed):
    """Neutral long-form: one row per team-game, with unified stat names and a win/loss flag."""
    # Winner side
    w_cols = {
        "WTeamID":"TeamID","WScore":"Score","WFGM":"FGM","WFGA":"FGA","WFGM3":"FG3M","WFGA3":"FG3A",
        "WFTM":"FTM","WFTA":"FTA","WOR":"OR","WDR":"DR","WAst":"Ast","WTO":"TO","WStl":"Stl","WBlk":"Blk","WPF":"PF"
    }
    w = rs_detailed.rename(columns=w_cols)
    w = w[["Season","TeamID","Score","FGM","FGA","FG3M","FG3A","FTM","FTA","OR","DR","Ast","TO","Stl","Blk","PF"]].copy()
    w["Result"] = 1  # win

    # Loser side
    l_cols = {
        "LTeamID":"TeamID","LScore":"Score","LFGM":"FGM","LFGA":"FGA","LFGM3":"FG3M","LFGA3":"FG3A",
        "LFTM":"FTM","LFTA":"FTA","LOR":"OR","LDR":"DR","LAst":"Ast","LTO":"TO","LStl":"Stl","LBlk":"Blk","LPF":"PF"
    }
    l = rs_detailed.rename(columns=l_cols)
    l = l[["Season","TeamID","Score","FGM","FGA","FG3M","FG3A","FTM","FTA","OR","DR","Ast","TO","Stl","Blk","PF"]].copy()
    l["Result"] = 0  # loss

    long_df = pd.concat([w, l], ignore_index=True)

    # Ensure numeric types (guard against any stray non-numerics)
    num_cols = ["Score","FGM","FGA","FG3M","FG3A","FTM","FTA","OR","DR","Ast","TO","Stl","Blk","PF"]
    long_df[num_cols] = long_df[num_cols].apply(pd.to_numeric, errors="coerce")
    return long_df

def aggregate_team_stats(long_df):
    """
    Aggregate averages by Season x Result (win/loss).
    Compute shooting percentages after aggregating totals to avoid tuple division in .agg.
    """
    agg = (long_df
           .groupby(["Season","Result"])
           .agg(
               Pts_avg   = ("Score","mean"),
               FGM_sum   = ("FGM","sum"),
               FGA_sum   = ("FGA","sum"),
               FG3M_sum  = ("FG3M","sum"),
               FG3A_sum  = ("FG3A","sum"),
               FTM_sum   = ("FTM","sum"),
               FTA_sum   = ("FTA","sum"),
               OR_avg    = ("OR","mean"),
               DR_avg    = ("DR","mean"),
               Ast_avg   = ("Ast","mean"),
               TO_avg    = ("TO","mean"),
               Stl_avg   = ("Stl","mean"),
               Blk_avg   = ("Blk","mean"),
               PF_avg    = ("PF","mean"),
               Games     = ("Score","count")
           )
           .reset_index()
          )

    # Safe division for percentages
    def safe_div(a, b):
        return np.where(b==0, np.nan, a / b)

    agg["FG_pct"]  = safe_div(agg["FGM_sum"],  agg["FGA_sum"])
    agg["FG3_pct"] = safe_div(agg["FG3M_sum"], agg["FG3A_sum"])
    agg["FT_pct"]  = safe_div(agg["FTM_sum"],  agg["FTA_sum"])

    return agg

# Build long forms
men_long = build_long_form(df_men_rs_detailed)
women_long = build_long_form(df_women_rs_detailed)

# Aggregate: average stats for winning vs losing teams
men_winloss_stats = aggregate_team_stats(men_long)
women_winloss_stats = aggregate_team_stats(women_long)

display(men_winloss_stats.head())
display(women_winloss_stats.head())
```

```python
def winner_loser_diff(agg_df, title_prefix=""):
    """
    Compute (Winner - Loser) season-averaged differences across seasons,
    then mean those season-level diffs to get an overall picture.
    """
    # Pivot to columns: Result=0 (loss), 1 (win)
    wide = agg_df.pivot_table(index="Season", columns="Result", values=[
        "Pts_avg","FG_pct","FG3_pct","FT_pct","OR_avg","DR_avg","Ast_avg","TO_avg","Stl_avg","Blk_avg","PF_avg"
    ])
    # Some seasons might be missing one side; drop such seasons
    wide = wide.dropna()

    # Compute per-season diff (win - loss) for each metric, then average across seasons
    diffs = {}
    for metric in ["Pts_avg","FG_pct","FG3_pct","FT_pct","OR_avg","DR_avg","Ast_avg","TO_avg","Stl_avg","Blk_avg","PF_avg"]:
        win_series  = wide[(metric, 1)]
        loss_series = wide[(metric, 0)]
        diffs[metric] = (win_series - loss_series).mean()

    diffs = pd.Series(diffs).sort_values(ascending=False).round(4)

    plt.figure(figsize=(8,5))
    sns.barplot(x=diffs.values, y=diffs.index)
    plt.title(f"{title_prefix} Which Stats Differentiate Winners?")
    plt.xlabel("Winner minus Loser (per-game or rate)")
    plt.show()

winner_loser_diff(men_winloss_stats, "Men — Regular Season")
winner_loser_diff(women_winloss_stats, "Women — Regular Season")
```

* For both men and women teams the following features are an indication on winning:
    * Points scored (Pts_avg)
    * Defensive rebounds (DR_avg)
    * Assists (Ast_avg)
    * Steals (Stl_avg)
    * Blocks (Blk_avg)
* For women teams the following feature are an indication on winning:
    * Offensive Rebounds (OR_avg)
* For men and women teams the following features are an indication on losing:
    * Turnovers (TO_avg)
    * Personal Fouls (PF_avg)

### Seed and Outcomes

```python
def add_numeric_seed(seeds_df):
    out = seeds_df.copy()
    out["SeedNum"] = out["Seed"].str[1:3].astype(int)
    return out

def merge_seeds_into_tourney(tourney_detailed, seeds_df):
    seeds_num = add_numeric_seed(seeds_df)
    left = tourney_detailed.merge(
        seeds_num.rename(columns={"TeamID":"WTeamID","SeedNum":"WSeed"}), on=["Season","WTeamID"], how="left"
    )
    both = left.merge(
        seeds_num.rename(columns={"TeamID":"LTeamID","SeedNum":"LSeed"}), on=["Season","LTeamID"], how="left"
    )
    both["SeedDiff"] = both["LSeed"] - both["WSeed"]  # positive => favored (better seed) won
    return both

men_tourney_seeds = merge_seeds_into_tourney(df_men_ncaa_detailed, df_men_ncaa_seeds)
women_tourney_seeds = merge_seeds_into_tourney(df_women_ncaa_detailed, df_women_ncaa_seeds)

def plot_seeddiff_hist(tourney_with_seeds, title_prefix=""):
    plt.figure(figsize=(8,5))
    sns.histplot(tourney_with_seeds["SeedDiff"].dropna(), bins=15)
    plt.title(f"{title_prefix}Seed Difference in NCAA Wins (+ means better seed won)")
    plt.xlim(-15,15)
    plt.xlabel("LSeed - WSeed")
    plt.show()

plot_seeddiff_hist(men_tourney_seeds, "Men — ")
plot_seeddiff_hist(women_tourney_seeds, "Women — ")
```

* In general higher seeded teams win games
* For women there are only few upsets --> highest seed difference = -9
* For men there are more upsets (more data also)

### Upset Rate

```python
def compute_upset_rates(tourney_with_seeds):
    """
    Compute upset rates (lower-seeded team wins) for all seed matchups.
    Returns a dataframe with columns: SeedHigh, SeedLow, Games, Upsets, UpsetRate.
    """
    df = tourney_with_seeds.dropna(subset=["WSeed","LSeed"]).copy()

    # Identify which seed is higher (numerically worse) and lower (better)
    df["WinnerSeed"] = np.where(df["WSeed"] < df["LSeed"], df["WSeed"], df["LSeed"])
    df["LoserSeed"]  = np.where(df["WSeed"] < df["LSeed"], df["LSeed"], df["WSeed"])
    df["LowerSeedWon"] = (df["WSeed"] > df["LSeed"]).astype(int)  # upset flag

    summary = (df.groupby(["WinnerSeed","LoserSeed"])
                 .agg(Games=("LowerSeedWon","count"),
                      Upsets=("LowerSeedWon","sum"))
                 .reset_index())
    summary["UpsetRate"] = summary["Upsets"] / summary["Games"]
    return summary

men_upsets = compute_upset_rates(men_tourney_seeds)
women_upsets = compute_upset_rates(women_tourney_seeds)

def plot_upset_heatmap(summary, title_prefix=""):
    pivot = summary.pivot(index="LoserSeed", columns="WinnerSeed", values="UpsetRate")
    plt.figure(figsize=(10,8))
    sns.heatmap(pivot, annot=True, fmt=".0%", cmap="RdYlGn_r", cbar_kws={'label':'Upset Rate'})
    plt.title(f"{title_prefix}Upset Rates by Seed Matchup\n(lower seed wins %)")
    plt.xlabel("Better Seed")
    plt.ylabel("Worse Seed")
    plt.show()

# Run again
plot_upset_heatmap(men_upsets, "Men — ")
plot_upset_heatmap(women_upsets, "Women — ")
```

* Mens tournament has more upsets overall, but also 7 years more data
* Upset rates are highest when the seed difference is smallest
* Mens seed 4 has very high upset rate in general

```python
def plot_corr_with_margin(rs_detailed, title_prefix=""):
    # Add Margin if missing
    df = rs_detailed.copy()
    if "Margin" not in df.columns:
        df["Margin"] = df["WScore"] - df["LScore"]

    stat_cols = ['WFGM','WFGA','WFGM3','WFGA3','WFTM','WFTA',
                 'WOR','WDR','WAst','WTO','WStl','WBlk','WPF']

    corrs = df[stat_cols + ['Margin']].corr(numeric_only=True)['Margin'].drop("Margin").sort_values()

    plt.figure(figsize=(8,6))
    sns.barplot(x=corrs.values, y=corrs.index, palette="coolwarm", hue=corrs.index, legend=False)
    plt.axvline(0, color="black", linewidth=1)
    plt.title(f"{title_prefix}Correlation of Stats with Win Margin")
    plt.xlabel("Pearson Correlation with Margin")
    plt.ylabel("Stat")
    plt.show()

    return corrs

# Men
men_corrs = plot_corr_with_margin(df_men_rs_detailed, "Men — ")

# Women
women_corrs = plot_corr_with_margin(df_women_rs_detailed, "Women — ")
```

* Field goal made (FGM) and assists (Ast) have the highest positive correlation with win margin for both men and women
* Turnovers (TO) and personal fouls (PF) have the highest negative correlation with win margin for both men and women

* Interestingly, both free throws made (FTM) and free throws attempted (FTA) have a negative correlation with win margin for both men and women
* All ohter stats have a positive correlation with win margin for both men and women with various degree of correlation
