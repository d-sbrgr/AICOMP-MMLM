```python
import pandas as pd
from itertools import combinations

from src.utils import get_data_directory
from src.submissions import generate_matchups
```

```python
YEAR = 2024
```

#### Load Team Data

```python
df_teams_men = pd.read_csv(get_data_directory() / "MTeams.csv")
df_teams_women = pd.read_csv(get_data_directory() / "WTeams.csv")
```

```python
df_teams_men.head()
```

```python
df_teams_women.head()
```

```python
# Only keep teams that are in D1 for the given year
df_teams_men = df_teams_men[(df_teams_men['FirstD1Season'] <= YEAR) & (df_teams_men['LastD1Season'] >= YEAR)]

# drop unused columns
df_teams_men = df_teams_men[['TeamID',]]
df_teams_women = df_teams_women[['TeamID',]]
df_teams_men.head()
```

```python
def create_matchup_df(df_teams):
    return pd.DataFrame(
        [(min(a, b), max(a, b)) for a, b in combinations(df_teams['TeamID'], 2)],
        columns=['lowerID', 'higherID']
    )
```

```python
df_matchups_men = create_matchup_df(df_teams_men)
df_matchups_women = create_matchup_df(df_teams_women)
```

```python
df_matchups_men.tail()
```

```python
df_matchups_women.tail()
```

```python
df_matchups = pd.concat((df_matchups_men, df_matchups_women)).reset_index(drop=True)
```

```python
df_matchups.head()
```

```python
# add new column 'ID' in the format 'YEAR_lowerID_higherID'
df_matchups['ID'] = df_matchups.apply(lambda row: f"{YEAR}_{row['lowerID']}_{row['higherID']}", axis=1)
```
