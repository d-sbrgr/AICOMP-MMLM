# Logistic Regression Model Experimentation

This notebook demonstrates how to train and evaluate the Logistic Regression model for March Madness predictions.

Logistic Regression provides fast, interpretable probabilistic predictions with built-in regularization.

```python
import sys

sys.path.append("..")

from src.dataloaders.simple import SeasonAverageDataLoader
from src.experiments import DefaultTracker, ExperimentConfig
from src.models.logistic_regression import LogisticRegressionHyperparamConfig, LogisticRegressionModel
```

## Configure Logistic Regression Hyperparameters

Key hyperparameters:

- **penalty**: Type of regularization ('l1', 'l2', 'elasticnet', None)
- **C**: Inverse of regularization strength (smaller = stronger regularization)
- **solver**: Optimization algorithm ('lbfgs', 'liblinear', 'saga', etc.)
- **max_iter**: Maximum number of iterations for convergence
- **l1_ratio**: ElasticNet mixing parameter (only for elasticnet penalty)

```python
# Example configuration with L2 regularization (Ridge)
lr_params = LogisticRegressionHyperparamConfig(penalty="l2", C=1.0, solver="lbfgs", max_iter=1000, tol=1e-4)

lr_params
```

## Setup Data Loader and Experiment Tracking

```python
# Setup data loader
data_loader = SeasonAverageDataLoader(num_features=20)

# Setup experiment tracking
experiment_config = ExperimentConfig(name="logistic-regression")
tracker = DefaultTracker(experiment_config)
```

## Train and Evaluate Logistic Regression Model

```python
# Create model instance
model = LogisticRegressionModel(data=data_loader, params=lr_params, cv=None, tracker=tracker)

model.fit(season=2024, start_season=2003)

model.validate()
```
