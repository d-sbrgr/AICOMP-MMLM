```python
import pandas as pd
```

```python
leaderboard = pd.read_csv("./../data/Kaggle_Leaderboard.csv")
```

```python
leaderboard = leaderboard.drop(
    columns=["TeamId", "TeamName", "SubmissionCount", "TeamMemberUserNames", "LastSubmissionDate"]
)
```

```python
leaderboard
```

```python
def get_rank(score: float, leaderboard: pd.DataFrame) -> int:
    return int(leaderboard[leaderboard.Score < score].iloc[-1].Rank + 1)
```

```python
get_rank(0.1179, leaderboard)
```
