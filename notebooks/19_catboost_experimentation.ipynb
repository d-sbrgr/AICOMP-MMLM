# Logistic Regression Model Experimentation

This notebook demonstrates how to train and evaluate the Logistic Regression model for March Madness predictions.

Logistic Regression provides fast, interpretable probabilistic predictions with built-in regularization.

```python
import sys

sys.path.append("..")

import pandas as pd
import numpy as np
from src.models.catboost import CatBoostModel, CatBoostHyperparamConfig
from src.dataloaders.simple import SeasonAverageDataLoader
from src.experiments import DefaultTracker, ExperimentConfig
from src.models.cross_validation.cv_config import CrossValidationConfig
```

## Configure CatBoost Hyperparameters

```python
# Example configuration with L2 regularization (Ridge)
lr_params = CatBoostHyperparamConfig(learning_rate=0.1, depth=6, l2_leaf_reg=3, iterations=1000)
lr_params
```

## Setup Data Loader and Experiment Tracking

```python
data_loader = SeasonAverageDataLoader(num_features=20)

experiment_config = ExperimentConfig(name="catboost")
tracker = DefaultTracker(experiment_config)
```

## Train and Evaluate Logistic Regression Model

```python
# Create model instance
model = CatBoostModel(data=data_loader, params=lr_params, cv=None, tracker=tracker)

model.fit(season=2024, start_season=2003)

model.validate()
```
