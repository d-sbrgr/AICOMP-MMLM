## Example - Use Quality in Data
An example of how the quality feature can be used in matchup data.

```python
from src.datasets.datasets import compact_tourney_results
from src.features.quality import apply_quality_to_matchups
from src.utils.constants import Columns
```

```python
m_qualities = pd.read_csv(get_data_directory() / f"M{TEAM_QUALITY_SEASONAL}.csv")
w_qualities = pd.read_csv(get_data_directory() / f"W{TEAM_QUALITY_SEASONAL}.csv")
```

```python
tourneys = compact_tourney_results()
```

```python
tourneys = apply_quality_to_matchups(
    tourneys, [Columns.WTEAM_ID, Columns.LTEAM_ID], pd.concat([m_qualities, w_qualities])
)
```
