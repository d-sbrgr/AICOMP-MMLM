# SVM Model - Experiments

This notebook demonstrates how to train and evaluate the Support Vector Machine (SVM) model for March Madness predictions.

The SVM model is implemented following the same structure as XGBoost and Random Forest models.

```python
import sys

sys.path.append("..")

from src.models.svm import SVMRegressorModel, SVMHyperparamConfig
from src.dataloaders.simple import SeasonAverageDataLoader
from src.experiments import DefaultTracker, ExperimentConfig
from src.models.cross_validation.cv_config import CrossValidationConfig
```

## Hyperparameters

```python
# Example configuration with RBF kernel (default)
svm_params = SVMHyperparamConfig(
    kernel="rbf",
    C=1.0,
    gamma="scale",
    epsilon=0.1,
    max_iter=1000,
    cache_size=500,  # Increase cache size for better performance
)

svm_params
```

## Experiment Setup

```python
# Setup data loader
data_loader = SeasonAverageDataLoader(num_features=20)

# Setup experiment tracking
experiment_config = ExperimentConfig(project="svm", name="svm")
tracker = DefaultTracker(experiment_config)
```

## Train and Evaluate SVM Model

```python
# Create model instance
model = SVMRegressorModel(data=data_loader, params=svm_params, cv=None, tracker=tracker)

model.fit(season=2024, start_season=2003)
model.validate()
```
